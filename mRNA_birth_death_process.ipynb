```python
#Let's test the steady state solution. The code above allow simulation of numer of production rates:

%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import odeint

def growth(state,t,a,b):
    X = state[0]
    Xd = a-b*X
    return [Xd]

Tmax=10000 #simultion time
t = np.arange(0,Tmax)        #defines a vector from 0 to 300 every 1 (where solution will be calculated)
X0 = 0                       #assign initial condition to X (X0)

d1=0.001
k1=3
x = odeint(growth,X0,t,args=(k1,d1))  #solve ODE using a odeint solver

k1=3
d1=0.01
x1 = odeint(growth,X0,t,args=(k1,d1))  #solve ODE using a odeint solver

k1=3
d1=0.1
x2 = odeint(growth,X0,t,args=(k1,d1))  #solve ODE using a odeint solver


#Plotting
plt.show()
plt.plot(t, x,'k',label="d1=0.001".format(3))  #numerical
plt.plot(t, x1,'b',label="d1=0.01".format(3))  #numerical
plt.plot(t, x2,'r',label="d1=0.1".format(3))  #numerical


plt.ylabel('mRNA', fontsize= 12)
plt.xlabel('time', fontsize= 12)
plt.legend(loc="upper right",ncol=1, title="Legend", fancybox=False)

```
