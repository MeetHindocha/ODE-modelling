```python
#First need to upload some generic packages
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import odeint


#write a function containing ODEs
def expression(x,t):
    m = x[0]       #mRNA level
    p = x[1]       #protein level


    k1 = 3        # mRNA/min
    k2 = 10       # protein/min/mRNA
    d1 = 0.02     # 1/min, corresponding to half-life of approximately 30 min t1/2=log(2)/d1
    d2 =  0.006   # 1/min, corresponding to half-life of approximately 120 min 

    md = k1-d1*m     #Equations
    pd = k2*m-d2*p
    return [md,pd]


t=np.linspace(0,1000,10000)  # times at which calculate the solution
x0 = [100,100000]                   # initial conditions of mrna and protein, [mrna_concentration, protein_concentration]
x01 = [0,0]                   # initial conditions
x02 = [400,500000]                   # initial conditions
x03 = [0,200000]
x = odeint(expression,x0,t)  # solve ODE using numerical solver
x1 = odeint(expression,x01,t)  # solve ODE using numerical solver
x2 = odeint(expression,x02,t)  # solve ODE using numerical solver
x3 = odeint(expression,x03,t)  # solve ODE using numerical solver


#Plotting now

plt.figure(1)
plt.subplot(211)
plt.plot(t, x[:,0], 'k-', label="x0=[100,100000]".format(3))  #plot mRNA
plt.plot(t, x1[:,0], 'b-', label="x0=[0,0]".format(3))  #plot mRNA
plt.plot(t, x2[:,0], 'r-', label="x0=[400,500000]".format(3))  #plot mRNA
plt.plot(t, x3[:,0], 'g-', label="x0=[0,200000]".format(3))  #plot mRNA
plt.xlabel('Time [mins]', fontsize=12)
plt.ylabel('mRNA number', fontsize= 12)
plt.legend(loc="upper right",ncol=1, title="Legend", fancybox=False)

plt.subplot(212)
plt.plot(t, x[:,1], 'k-', linewidth=2)   #plot protein
plt.plot(t, x1[:,1], 'b-', linewidth=2)   #plot protein
plt.plot(t, x2[:,1], 'r-', linewidth=2)   #plot protein
plt.plot(t, x3[:,1], 'g-', linewidth=2)   #plot protein
plt.xlabel('Time [mins]', fontsize=12)
plt.ylabel('Protein number', fontsize= 12)

plt.show()
```

```python
# numerical solution
#Lets simulate this system in Python

#First need to upload some generic packages
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import odeint

def autoactivation(x,t):
    m = x[0]
    p = x[1]
    
    K=50000
    n=25
    k1 = 3        #mRNA/min
    k2 = 10       #protein/min/mRNA
    d1 = 0.02     #1/min, corresponding to half-life of approximately 30 min 
    d2 =  0.006   #1/min, corresponding to half-life of approximately 120 min 
    
    md = k1*p**n/(K**n+p**n)-d1*m
    pd = k2*m-d2*p

    return [md,pd]


t=np.linspace(0,5000,1000)
x0 = [100,100000]
x = odeint(autoactivation,x0,t)

plt.figure(1)
plt.subplot(211)
plt.plot(t, x[:,0], '-', linewidth=2)
plt.xlabel('Time [mins]', fontsize=12)
plt.ylabel('mRNA number', fontsize= 12)

plt.subplot(212)
plt.plot(t, x[:,1], '-', linewidth=2)
plt.xlabel('Time [mins]', fontsize=12)
plt.ylabel('Protein number', fontsize= 12)

plt.show()
```
